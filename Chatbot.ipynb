## Conversational AI(chatbot)

```python
# imports

import os
from dotenv import load_dotenv
from openai import OpenAI
import gradio as gr
```

```python
openai = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')
```

```python
system_message = "You are a helpful assistant"
```

```python
## Gradio itself converting the hstory in to OPENAI format
```

```python
## So combine system message, History and message.
def chat(message, history):
    messages = [{"role": "system", "content": system_message}] + history + [{"role": "user", "content": message}]

    print("History is:")
    print(history)
    print("And messages is:")
    print(messages)

    stream = openai.chat.completions.create(model="llama3.2", messages=messages, stream=True)

    response = ""
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        yield response
```

### Here comes the Gradio UI 

```python
gr.ChatInterface(fn=chat, type="messages").launch()
```
